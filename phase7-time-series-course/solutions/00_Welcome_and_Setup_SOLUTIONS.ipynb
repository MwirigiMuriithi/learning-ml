### ✅ Solution 0.2 · Which crop is the most expensive?

**Problem recap:** Using the full table `market` (all four commodities stacked), compute the **average price per commodity** and
  print them from highest to lowest.

```python
avg = market.groupby('commodity')['price'].mean().sort_values(ascending=False)
print(avg.round(1))
```

**Why this works:** `groupby` splits the table into one piece per commodity, applies `mean`, and combines the results. Rice is the priciest per kg in our simulated market.

### ✅ Solution 0.3 · Guess before you compute

**Problem recap:** Look at the maize plot above. In **plain words** (write it in a markdown cell or a comment): which months do you think maize is
  cheapest and which are most expensive? Then verify with code: compute the average price for each calendar month of the year
  *after removing the long-run trend* by dividing each price by its 52-week rolling mean.

```python
trend = maize['price'].rolling(52, center=True).mean()
ratio = maize['price'] / trend
print(ratio.groupby(ratio.index.month).mean().round(3))
```

**Why this works:** Dividing by a moving average removes the slow trend so the seasonal pattern (harvest lows, lean-season highs) pops out. This is a preview of decomposition in Lesson 1.
