### 🧭 How every lesson is organised

Every notebook follows the same rhythm, like a good lecture:

1. **🎯 Learning objectives**: what you'll be able to do afterwards.
2. **📖 Lecture**: ideas explained with analogies, pictures, and live code.
3. **🧑‍🏫 Worked examples**: I solve a problem *out loud*, showing my reasoning, including the wrong turns.
4. **🤔 Pause & Predict**: you guess *before* running the code. (Guessing first is how your brain actually learns.)
5. **🧩 Problems**: you solve them. Each has **hints** (open one at a time) and an automatic **self-check**.
6. **🎟️ Exit ticket**: five quick questions to see if the ideas stuck.
7. **📌 Recap**: one-page summary to screenshot and keep.
