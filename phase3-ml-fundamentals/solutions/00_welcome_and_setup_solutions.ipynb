### 🧩 Warm-up 0.1: The average student &nbsp; ⭐

Using the `students` DataFrame loaded above, compute the **average exam score** and store it in a variable called `avg_score`.

*(We will use the `students` dataset again and again. It records study habits and exam results for 200 students.)*

```python
students = load("students.csv")

avg_score = students["exam_score"].mean()
print(avg_score)
```

```python
# ✔ Self-check
check("avg_score is a number near 56", lambda: abs(avg_score - 56.2) < 0.5)
```

**🧑‍🏫 Walkthrough.** `students["exam_score"]` picks one column (a pandas *Series*), and `.mean()` averages it. This is the simplest "model" you will ever meet: if someone asks *"what score will a random student get?"*, answering **56.2** is a legitimate prediction. Hold on to that idea: in Lesson 6 we'll call it the **baseline**, and every real model has to beat it.
