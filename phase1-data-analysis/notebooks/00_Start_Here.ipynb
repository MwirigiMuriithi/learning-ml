## 🐍 Step 2: A 15-minute Python refresher

You need *only* these ideas. If anything here is unfamiliar, spend time on it now; it will pay for itself in every lesson.

### Variables, numbers, text, f-strings

```python
price = 52.5                       # float
n_markets = 9                      # int
crop = "Maize"                     # str
is_wholesale = True                # bool

print(f"{crop} costs {price} KES/kg in {n_markets} markets")      # f-string: put values inside text
print(f"A 10% rise would give {price * 1.10:.2f}")                # :.2f = 2 decimal places
print(type(price), type(n_markets), type(crop), type(is_wholesale))
```

### Lists and dictionaries

```python
crops = ["Maize", "Beans", "Kale"]                 # list: ordered, index from 0
crops.append("Onions")
print(crops[0], crops[-1], len(crops), crops[1:3])

base = {"Maize": 48, "Beans": 125, "Kale": 32}     # dict: key -> value
base["Onions"] = 72
print(base["Beans"], list(base.keys()), base.get("Rice", "not sold"))
for crop, p in base.items():
    print(f"  {crop:<7} {p}")
```

### Loops, conditions, comprehensions

```python
for crop in crops:
    if crop == "Maize":
        print("staple:", crop)
    elif len(crop) > 4:
        print("long name:", crop)
    else:
        print("other:", crop)

squares = [x ** 2 for x in range(1, 6)]                         # list comprehension: a loop in one line
dear = [c for c, p in base.items() if p > 50]                   # ...with a filter
inflated = {c: round(p * 1.07, 1) for c, p in base.items()}     # dict comprehension
print(squares, dear, inflated)
```

### Functions and lambdas

```python
def to_usd(kes, rate=129.0):
    """Convert KES to USD. (The text in triple quotes is the docstring: what the function does.)"""
    return kes / rate

print(round(to_usd(1290), 2), round(to_usd(1290, rate=130), 2))

double = lambda x: x * 2            # a tiny unnamed function; you'll meet these inside pandas' apply/transform
print(double(21), sorted(crops, key=lambda c: len(c)))
```

### Reading error messages (a superpower)

Errors are *information*. Read **from the bottom up**: the last line names the problem; the lines above show where.

| You see… | It usually means… | Try… |
|---|---|---|
| `KeyError: 'Price'` | column/key doesn't exist (spelling? capitals? spaces?) | `df.columns` |
| `ValueError: operands could not be broadcast together` | array shapes don't line up | print both `.shape`s |
| `TypeError: unsupported operand type(s)` | mixing text and numbers | `df.dtypes`, `.astype(float)` |
| `AttributeError: 'Series' object has no attribute …` | wrong object type / typo in a method | `type(x)`, `dir(x)` |
| `SettingWithCopyWarning` | you edited a copy by accident | use `.loc[mask, col] = value` or `.copy()` |
| `FileNotFoundError` | wrong path | `DATA_DIR / "file.csv"` |
| `NameError: name 'df' is not defined` | cell not run (or kernel restarted) | run Setup cell first |
